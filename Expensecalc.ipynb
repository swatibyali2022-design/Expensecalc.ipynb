```python
# ============================================================
# 💰 AI EXPENSE TRACKER - GOOGLE COLAB
# Uses ROLE PROMPTING
# ============================================================

# Install OpenAI library
!pip -q install openai


# Import libraries
from openai import OpenAI
import getpass


# ============================================================
# 1. ENTER API KEY
# ============================================================

print("🔐 Enter your OpenAI API Key")
api_key = getpass.getpass("API Key: ")

client = OpenAI(api_key=api_key)


# ============================================================
# 2. EXPENSE LIST
# ============================================================

expenses = []


# ============================================================
# 3. ADD EXPENSE
# ============================================================

def add_expense():

    print("\n========== ADD EXPENSE ==========")

    name = input("Enter expense name: ")

    while True:
        try:
            amount = float(input("Enter amount (₹): "))

            if amount <= 0:
                print("Amount must be greater than 0.")
            else:
                break

        except ValueError:
            print("Please enter a valid amount.")

    category = input("Enter category: ")

    expense = {
        "name": name,
        "amount": amount,
        "category": category
    }

    expenses.append(expense)

    print("\n✅ Expense added successfully!")


# ============================================================
# 4. VIEW EXPENSES
# ============================================================

def view_expenses():

    print("\n========== YOUR EXPENSES ==========")

    if len(expenses) == 0:
        print("No expenses added yet.")
        return

    for i, expense in enumerate(expenses, 1):

        print(
            f"{i}. {expense['name']} | "
            f"₹{expense['amount']:.2f} | "
            f"{expense['category']}"
        )


# ============================================================
# 5. CALCULATE TOTAL
# ============================================================

def calculate_total():

    if len(expenses) == 0:
        print("\nNo expenses available.")
        return

    total = sum(expense["amount"] for expense in expenses)

    print("\n========== TOTAL ==========")
    print(f"💰 Total Spending: ₹{total:.2f}")


# ============================================================
# 6. AI ANALYSIS
# ============================================================

def ai_analysis():

    if len(expenses) == 0:
        print("\n❌ Add some expenses before using AI analysis.")
        return

    # Convert expenses into text
    expense_data = ""

    for expense in expenses:

        expense_data += (
            f"Expense: {expense['name']}, "
            f"Amount: ₹{expense['amount']:.2f}, "
            f"Category: {expense['category']}\n"
        )


    # ========================================================
    # ROLE PROMPTING
    # ========================================================

    role_prompt = """
You are SpendSense AI, a friendly and practical personal
money coach.

Your role is to analyze a user's expense data and help them
understand their spending habits.

Your responsibilities are:

1. Calculate the total amount spent.
2. Identify the highest spending category.
3. Identify unusually large expenses if there are any.
4. Explain the user's spending pattern.
5. Give 3 practical money-saving suggestions.
6. Give one simple weekly money-saving challenge.

Important rules:

- Use ONLY the expense data provided by the user.
- Do not invent or assume expenses.
- Do not criticize or judge the user.
- Keep the explanation simple.
- Use Indian Rupees (₹).
- Make the suggestions practical.

Return the answer using exactly these sections:

💰 SPENDING SUMMARY

📊 HIGHEST SPENDING CATEGORY

🔎 SPENDING PATTERN

⚠️ LARGE EXPENSES

💡 MONEY-SAVING SUGGESTIONS

🎯 THIS WEEK'S CHALLENGE
"""


    # User data prompt
    user_prompt = f"""
Analyze the following expense data:

{expense_data}
"""


    print("\n🤖 SpendSense AI is analyzing your expenses...")
    print("Please wait...\n")


    # Send request to AI
    response = client.responses.create(
        model="gpt-5.6-luna",
        instructions=role_prompt,
        input=user_prompt
    )


    # Display AI response
    print(response.output_text)


# ============================================================
# 7. DELETE EXPENSE
# ============================================================

def delete_expense():

    if len(expenses) == 0:
        print("\n❌ No expenses to delete.")
        return

    view_expenses()

    try:
        number = int(input("\nEnter expense number to delete: "))

        if 1 <= number <= len(expenses):

            removed = expenses.pop(number - 1)

            print(
                f"\n🗑️ Deleted: {removed['name']} "
                f"(₹{removed['amount']:.2f})"
            )

        else:
            print("❌ Invalid expense number.")

    except ValueError:
        print("❌ Please enter a valid number.")


# ============================================================
# 8. MAIN PROGRAM
# ============================================================

def main():

    while True:

        print("\n")
        print("=" * 45)
        print("          💰 SPENDSENSE AI")
        print("       AI EXPENSE TRACKER")
        print("=" * 45)

        print("1. ➕ Add Expense")
        print("2. 📋 View Expenses")
        print("3. 💰 Calculate Total")
        print("4. 🤖 AI Money Coach")
        print("5. 🗑️ Delete Expense")
        print("6. 🚪 Exit")

        print("=" * 45)

        choice = input("Enter your choice: ")


        if choice == "1":

            add_expense()


        elif choice == "2":

            view_expenses()


        elif choice == "3":

            calculate_total()


        elif choice == "4":

            ai_analysis()


        elif choice == "5":

            delete_expense()


        elif choice == "6":

            print("\n👋 Thank you for using SpendSense AI!")
            break


        else:

            print("\n❌ Invalid choice. Please select 1-6.")


# ============================================================
# START PROGRAM
# ============================================================

main()
```
